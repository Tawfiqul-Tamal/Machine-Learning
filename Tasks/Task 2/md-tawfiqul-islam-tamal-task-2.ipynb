# Univariate analysis

# 1. Barplot

```python
# Average Age grouped by Education level (Employee dataset)
plt.figure(figsize=(8,5))
sns.barplot(data=employee, x='Education', y='Age', hue='Education', palette='viridis', legend=False)
plt.title('Average Age by Education Level')
plt.xlabel('Education')
plt.ylabel('Average Age')
plt.show()
```

# 2. Piechart

```python
# Distribution of restaurant visits by Day (Tips dataset)
day_counts = tips['day'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(day_counts.values, labels=day_counts.index, autopct='%1.1f%%',
        colors=sns.color_palette('pastel'), startangle=90)
plt.title('Proportion of Bills by Day')
plt.show()

# Seaborn doesn't have a built-in pie chart, so matplotlib.pyplot.pie is used here,
# with a seaborn color palette applied to keep the look consistent.
```

# 3. Countplot

```python
# Count of patients by DEATH_EVENT outcome (Heart Failure dataset)
plt.figure(figsize=(6,5))
sns.countplot(data=heart, x='DEATH_EVENT', hue='DEATH_EVENT', palette='Set2', legend=False)
plt.title('Count of Patients by Death Event (0 = Survived, 1 = Deceased)')
plt.xlabel('Death Event')
plt.ylabel('Number of Patients')
plt.show()
```

# 4. Histogram

```python
# Distribution of patient Age (Heart Failure dataset)
plt.figure(figsize=(8,5))
sns.histplot(data=heart, x='age', bins=20, color='steelblue')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

# 5. Kde plot

```python
# Density estimate of Total Bill (Tips dataset)
plt.figure(figsize=(8,5))
sns.kdeplot(data=tips, x='total_bill', fill=True, color='darkorange')
plt.title('KDE of Total Bill Amount')
plt.xlabel('Total Bill ($)')
plt.show()

# KDE smooths a histogram into a continuous curve, making the overall
# shape of the distribution easier to read.
```

# 6. Boxplot

```python
# Spread of Age values (Employee dataset)
plt.figure(figsize=(6,5))
sns.boxplot(data=employee, y='Age', color='mediumseagreen')
plt.title('Boxplot of Employee Age')
plt.ylabel('Age')
plt.show()

# The box marks the interquartile range (IQR), the line inside is the median,
# and dots beyond the whiskers are potential outliers.
```

# Bivariate analysis

# 1. Scatter plot

```python
# Relationship between Total Bill and Tip amount (Tips dataset)
plt.figure(figsize=(8,5))
sns.scatterplot(data=tips, x='total_bill', y='tip', color='crimson')
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

# 2. Rugplot

```python
# Rugplot layered on a histogram of Age (Heart Failure dataset)
plt.figure(figsize=(8,3))
sns.histplot(data=heart, x='age', bins=20, color='lightgray')
sns.rugplot(data=heart, x='age', color='black')
plt.title('Age Distribution with Rugplot (each tick = one patient)')
plt.xlabel('Age')
plt.show()
```

# 3. Regplot

```python
# Total Bill vs Tip with a fitted regression line (Tips dataset)
plt.figure(figsize=(8,5))
sns.regplot(data=tips, x='total_bill', y='tip', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Total Bill vs Tip with Regression Line')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()

# regplot combines a scatterplot with a fitted linear regression line,
# giving a quick visual read on correlation strength and direction.
```

# 4. Joint plot

```python
# Joint distribution of Age and Serum Creatinine (Heart Failure dataset)
sns.jointplot(data=heart, x='age', y='serum_creatinine', kind='scatter', color='teal')
plt.suptitle('Age vs Serum Creatinine', y=1.02)
plt.show()

# jointplot puts a bivariate scatterplot in the center with univariate
# histograms on the margins, showing the relationship and each variable's spread together.
```

# 5. 2D-kde plot

```python
# 2D density of Total Bill vs Tip (Tips dataset)
plt.figure(figsize=(8,6))
sns.kdeplot(data=tips, x='total_bill', y='tip', fill=True, cmap='mako', thresh=0.05)
plt.title('2D KDE of Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

# 6. Box plot

```python
# Age spread grouped by Education level (Employee dataset)
plt.figure(figsize=(9,5))
sns.boxplot(data=employee, x='Education', y='Age', hue='Education', palette='coolwarm', legend=False)
plt.title('Age Distribution by Education Level')
plt.xlabel('Education')
plt.ylabel('Age')
plt.show()
```

# 7. Violin plot

```python
# Age distribution split by Death Event (Heart Failure dataset)
plt.figure(figsize=(8,5))
sns.violinplot(data=heart, x='DEATH_EVENT', y='age', hue='DEATH_EVENT', palette='muted', legend=False)
plt.title('Age Distribution by Death Event')
plt.xlabel('Death Event')
plt.ylabel('Age')
plt.show()

# A violin plot is like a boxplot but also traces the KDE shape of the data,
# so it can reveal multi-modal distributions that a plain boxplot would hide.
```

# 8. Swarm plot

```python
# Individual Tip values across Days (Tips dataset)
plt.figure(figsize=(8,5))
sns.swarmplot(data=tips, x='day', y='tip', hue='day', palette='Set1', legend=False, size=4)
plt.title('Tip Amount by Day (each dot = one bill)')
plt.xlabel('Day')
plt.ylabel('Tip ($)')
plt.show()
```

# 9. Heatmap

```python
# Correlation heatmap of numeric features (Heart Failure dataset)
plt.figure(figsize=(10,8))
corr = heart.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap - Heart Failure Dataset')
plt.show()

# Values near +1 or -1 mean strong correlation; values near 0 mean the two
# features have little to no linear relationship.
```

# Multivariate analysis

# 1. Scatter plot

```python
# Total Bill vs Tip, colored by Day and sized by party Size (Tips dataset)
plt.figure(figsize=(9,6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day', size='size',
                sizes=(30,200), palette='deep', alpha=0.8)
plt.title('Total Bill vs Tip (colored by Day, sized by Party Size)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.legend(bbox_to_anchor=(1.02,1), loc='upper left')
plt.tight_layout()
plt.show()

# Adding hue and size turns a simple 2D scatterplot into a multivariate view,
# since three extra dimensions (day, size, and the two axes) are shown at once.
```

# 2. Pair plot

```python
# Pairwise relationships among key numeric features, colored by Death Event (Heart Failure dataset)
cols = ['age','ejection_fraction','serum_creatinine','serum_sodium','platelets','DEATH_EVENT']
sns.pairplot(heart[cols], hue='DEATH_EVENT', palette='husl', diag_kind='kde')
plt.suptitle('Pairplot of Key Clinical Features', y=1.02)
plt.show()

# pairplot draws every pairwise scatterplot between the chosen numeric columns
# plus a KDE on the diagonal, giving a fast overview of multiple relationships at once.
```

# 3. Bubble chart

```python
# Age vs Experience in Current Domain, bubble size by Payment Tier (Employee dataset)
plt.figure(figsize=(9,6))
sns.scatterplot(data=employee, x='Age', y='ExperienceInCurrentDomain',
                size='PaymentTier', hue='PaymentTier', sizes=(40,300),
                palette='cool', alpha=0.7, legend='brief')
plt.title('Age vs Experience (bubble size = Payment Tier)')
plt.xlabel('Age')
plt.ylabel('Experience in Current Domain (years)')
plt.tight_layout()
plt.show()

# A bubble chart is just a scatterplot where a third numeric variable controls
# the size of each point, letting three variables be compared in one 2D plot.
```

# 4. Facet grid

```python
# Total Bill vs Tip split into a grid by Day and Time (Tips dataset)
g = sns.FacetGrid(tips, col='time', row='day', hue='sex', margin_titles=True, height=2.8)
g.map(sns.scatterplot, 'total_bill', 'tip', alpha=0.7)
g.add_legend()
g.fig.suptitle('Total Bill vs Tip, split by Day and Time', y=1.02)
plt.show()

# FacetGrid repeats the same plot across subsets of the data (here Day x Time),
# which is a quick way to spot whether a relationship changes across groups.
```

# 5. Pairgrid

```python
# Custom pairwise grid of numeric Employee features, colored by Gender
cols = ['Age','JoiningYear','PaymentTier','ExperienceInCurrentDomain']
g = sns.PairGrid(employee[cols + ['Gender']], hue='Gender', palette='Set1')
g.map_upper(sns.scatterplot, alpha=0.6)
g.map_lower(sns.kdeplot)
g.map_diag(sns.histplot)
g.add_legend()
plt.show()

# PairGrid gives full control over what goes on the diagonal versus the
# upper/lower triangles, unlike pairplot which uses one plot type throughout.
```

# 6. Joint plot

```python
# Total Bill vs Tip, split by Smoker status (Tips dataset)
sns.jointplot(data=tips, x='total_bill', y='tip', hue='smoker', kind='scatter', palette='Set2')
plt.suptitle('Total Bill vs Tip by Smoker Status', y=1.02)
plt.show()

# Adding hue='smoker' turns the earlier bivariate jointplot into a multivariate
# view, since it now compares the total_bill-tip relationship across two groups.
```

# 7. Boxen plot

```python
# Serum Creatinine levels grouped by Death Event and Sex (Heart Failure dataset)
plt.figure(figsize=(9,6))
sns.boxenplot(data=heart, x='DEATH_EVENT', y='serum_creatinine', hue='sex', palette='Spectral')
plt.title('Serum Creatinine by Death Event and Sex')
plt.xlabel('Death Event')
plt.ylabel('Serum Creatinine')
plt.legend(title='Sex (0=F, 1=M)')
plt.show()

# Boxenplot (a.k.a letter-value plot) extends the boxplot idea with more
# quantile levels, which works better than a normal boxplot on larger datasets.
```
